```python
# going to have to give the function two arguments 
def convertShelter(cd, borough):
    if cd < 10: 
        str_cd = str(cd)
        str_cd = "0" + str_cd

    else: 
        str_cd = str(cd)

    if borough == "Manhattan":
        str_cd_code = f"M{str_cd}"
    elif borough == "Brooklyn":
        str_cd_code = f"K{str_cd}"
    elif borough == "Bronx":
        str_cd_code = f"B{str_cd}"
    elif borough == "Staten Island":
        str_cd_code = f"S{str_cd}"
    elif borough == "Queens":
        str_cd_code = f"Q{str_cd}"
    else: str_cd_code = "FLAG"
    return str_cd_code

# it has to be indnted, so it knows to read the sentence above it 


# if borough = manhattan, make it X 
# turn community district into a two digit number 
    
```
